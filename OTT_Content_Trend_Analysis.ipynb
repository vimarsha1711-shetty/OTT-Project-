# OTT Streaming Content Trend Analysis
**Practice version — sample data only**

This beginner-friendly notebook explores a small practice dataset with platform, content type, genre, country and release year. The included records are generated examples, not official Netflix, Prime Video or Hotstar catalogue records. Replace the sample CSV with an authorized real dataset before making factual claims in a final submission.


```python
import pandas as pd
import matplotlib.pyplot as plt
from pathlib import Path

DATA_PATH = Path("../data/ott_titles_practice_sample.csv")
df = pd.read_csv(DATA_PATH)
df.head()
```

## 1. Understand the data

```python
print("Rows and columns:", df.shape)
print("\\nColumn names:", list(df.columns))
df.info()
```

## 2. Check missing values and duplicates

```python
print("Missing values by column:")
print(df.isna().sum())
print("\\nDuplicate rows:", df.duplicated().sum())
```

## 3. Basic cleaning

```python
df["release_year"] = pd.to_numeric(df["release_year"], errors="coerce")
for col in ["platform", "type", "title", "country", "rating", "genre", "duration"]:
    df[col] = df[col].fillna("Unknown")
df = df.drop_duplicates()
df["date_added"] = pd.to_datetime(df["date_added"], errors="coerce")
df["added_year"] = df["date_added"].dt.year
df.head()
```

## 4. Six visualizations

```python
# Chart 1: Number of titles by platform
df["platform"].value_counts().plot(kind="bar", title="Titles by Platform", xlabel="Platform", ylabel="Titles", rot=20)
plt.tight_layout()
plt.show()
```

```python
# Chart 2: Movies vs TV Shows
df["type"].value_counts().plot(kind="pie", autopct="%1.1f%%", title="Movies vs TV Shows")
plt.ylabel("")
plt.show()
```

```python
# Chart 3: Titles by genre
df["genre"].value_counts().sort_values().plot(kind="barh", title="Titles by Genre", xlabel="Titles")
plt.tight_layout()
plt.show()
```

```python
# Chart 4: Titles by release year
df.groupby("release_year").size().sort_index().plot(kind="line", marker="o", title="Titles by Release Year", xlabel="Release year", ylabel="Titles")
plt.tight_layout()
plt.show()
```

```python
# Chart 5: Most represented countries in this sample
df["country"].value_counts().head(8).plot(kind="bar", title="Top Countries in Sample", xlabel="Country", ylabel="Titles", rot=35)
plt.tight_layout()
plt.show()
```

```python
# Chart 6: Movies and TV shows by platform
pd.crosstab(df["platform"], df["type"]).plot(kind="bar", stacked=True, title="Content Type by Platform", ylabel="Titles", rot=20)
plt.tight_layout()
plt.show()
```

## 5. SQL-style business questions using Pandas

```python
# Q1. Which platform has the most titles?
print(df.groupby("platform").size().sort_values(ascending=False))

# Q2. Which genre appears most often?
print(df.groupby("genre").size().sort_values(ascending=False).head(5))

# Q3. What is the movie/TV split by platform?
print(pd.crosstab(df["platform"], df["type"]))

# Q4. Which countries appear most often?
print(df.groupby("country").size().sort_values(ascending=False).head(5))

# Q5. What is the release-year trend?
print(df.groupby("release_year").size().sort_index().tail(10))
```

## 6. Export cleaned data

```python
output_path = Path("../data/ott_titles_cleaned.csv")
df.to_csv(output_path, index=False)
print("Saved cleaned file to:", output_path)
```
